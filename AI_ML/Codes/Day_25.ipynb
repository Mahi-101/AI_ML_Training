# Simple Linear Regression

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
```

```python
data = pd.read_csv(r"E:\AI_ML_Training\AI_ML\raw_materials\Bengaluru_House_Price.csv")
data.head()
```

```python
data.dtypes
```

```python
data.isnull().sum()
```

```python
data_2 = data[['balcony', 'price']]
data_2.head()
```

```python
data_2 = data_2.dropna()
data_2.head()
```

```python
data_2.isnull().sum()
```

```python
x = data_2[['balcony']]
y = data_2['price']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
print(f"Shape of x_train : {x_train.shape}")
print(f"Shape of x_test : {x_test.shape}")
print(f"Shape of y_train : {y_train.shape}")
print(f"Shape of y_test : {y_test.shape}")
```

```python
model = LinearRegression()
```

```python
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
print(f" r2 score of model : {r2_score(y_test, y_pred)}")
```

```python
plt.figure(figsize=(7,5))

plt.scatter(x, y, color="blue", label="Actual Data")

plt.plot(x, model.predict(x),
         color="red",
         linewidth=2,
         label="Regression Line")

plt.xlabel("balcony ")
plt.ylabel("Price (Lakhs)")
plt.title("Simple Linear Regression")
plt.legend()

plt.show()
```

## Removing Outliers

```python
def remove_outlier_IQR(data_2, column):
    Q1 = data_2[column].quantile(0.25)

    Q3 = data_2[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5*IQR
    upper_bound = Q3 + 1.5*IQR

    print(f"{column} outlier boundaries :")
    print("Q1 : ", Q1)
    print("Q2 : ", Q3)
    print("IQR : ", IQR)
    print("lower_bound :", lower_bound)
    print("Upper_bound :", upper_bound)

    filtered_data = data_2[(data_2[column] >= lower_bound) & (data_2[column] <= upper_bound)]

    return filtered_data
```

```python
clean_train = train_df_num_col[['LotArea', 'SalePrice']]
clean_train.head()
```

```python
clean_train = remove_outlier_IQR(clean_train, 'LotArea')

print("After LotArea outlier removal")
clean_train.shape
```

```python
clean_train = remove_outlier_IQR(clean_train, 'SalePrice')

print("After SalePrice outlier removal")
clean_train.shape
```

```python
fig, axes = plt.subplots(1,2, figsize = (14, 5))

axes[0].boxplot(clean_train['LotArea'], vert = False, patch_artist = True)
axes[0].set_title("Total Sqaure Feet", fontsize = 12, fontweight = "bold")
axes[0].set_xlabel("Lot Area", fontsize = 11)
axes[0].grid(axis = 'x', linestyle = "--", alpha = 0.3)

axes[1].boxplot(clean_train['SalePrice'], vert = False, patch_artist = True)
axes[1].set_title("Price", fontsize = 12, fontweight = "bold")
axes[1].set_xlabel(" Sales price (rupees lakhs)", fontsize = 11)
axes[1].grid(axis = 'x', linestyle = "--", alpha = 0.3)

plt.suptitle("Box plot after outlier removal", fontsize = 16, fontweight = "bold")
plt.tight_layout()
```
