```python
missing_val_per = data.isnull().sum()/data.shape[0] * 100
missing_val_per
```

```python
missing_val_col_gre_20 = missing_val_per[missing_val_per > 20].keys()
missing_val_col_gre_20
```

```python
data_drop_col = data.drop(columns = missing_val_col_gre_20)
data_drop_col.shape
```

```python
data_drop_numeric = data_drop_col.select_dtypes(include = ['int64', 'float64'])
data_drop_numeric.head()
```

```python
plt.figure(figsize = (15,7))
sns.heatmap(data_drop_numeric.isnull())
```

```python
data_drop_numeric[data_drop_numeric.isnull().any(axis = 1)]
```

```python
missing_num_col_name = [var for var in data_drop_numeric.columns if data_drop_numeric[var].isnull().sum() > 0]
missing_num_col_name
```

```python
plt.figure(figsize= (15,7))
for i, var in enumerate(missing_num_col_name):
    plt.subplot(2,2,i+1)
    sns.distplot(data_drop_numeric[var], bins = 20, kde_kws = {'linewidth': 2, 'color': '#DC148C'})
```

```python
data_num_mean = data_drop_numeric.fillna(data_drop_numeric.mean())
data_num_mean.isnull().sum().sum()
```

```python
plt.figure(figsize= (20,15))
sns.set()
for i, var in enumerate(missing_num_col_name):
    plt.subplot(2,2,i+1)
    sns.distplot(data_drop_numeric[var], bins = 20, kde_kws = {'linewidth': 1, 'color': '#DC148C'}, label = "Original" )
    sns.distplot(data_num_mean[var], bins = 20, kde_kws = {'linewidth': 2, 'color': 'g'}, label = 'mean')
    plt.legend()
    
```

```python
data_num_median = data_drop_numeric.fillna(data_drop_numeric.median())
data_num_median.isnull().sum().sum()
```

```python
# From the below plots we can decide that which value we have to fill in the null places. the value that peak high select that value.
# Here we can see that the median peak is high so select that value to fill the null values.

plt.figure(figsize= (20,15))
sns.set()
for i, var in enumerate(missing_num_col_name):
    plt.subplot(2,2,i+1)
    sns.distplot(data_drop_numeric[var], bins = 20, kde_kws = {'linewidth': 1, 'color': '#DC148C'}, label = "Original" )
    sns.distplot(data_num_median[var], bins = 20, kde_kws = {'linewidth': 3, 'color': 'b'}, label = 'Meadian')
    sns.distplot(data_num_mean[var], bins = 20, kde_kws = {'linewidth': 2, 'color': 'g'}, label = 'Mean')
    plt.legend()
```

```python
for i, var in enumerate(missing_num_col_name):
    plt.figure(figsize= (20,15))
    plt.subplot(3,1,1)
    sns.boxplot(x = data[var], orient = 'h', label = 'Original Data')
    plt.subplot(3,1,2)
    sns.boxplot(x = data_num_mean[var], orient = 'h', label = 'Mean Data')
    plt.subplot(3,1,3)
    sns.boxplot(x = data_num_median[var], orient = 'h', label = 'Median Data')
```
