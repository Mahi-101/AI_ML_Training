# QQ Plot

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
```

```python
normal_data = np.random.normal(50, 10, 1000)

right_skewed = np.random.exponential(scale = 10, size = 1000)

left_skewed = -np.random.exponential(scale = 10, size = 1000) + 50
```

```python
plt.figure(figsize= (10,5))

plt.subplot(1, 3, 1)
stats.probplot(normal_data, dist = "norm", plot = plt)
plt.title("Normal Distribution")
```

```python
plt.subplot(1, 3, 2)
stats.probplot(right_skewed, dist = "norm", plot = plt)
plt.title(" Right Skewed Distribution (Positive Skewed)")
```

```python
plt.subplot(1, 3, 3)
stats.probplot(left_skewed, dist = "norm", plot = plt)
plt.title("Left Skewed Distribution (Negative Skewed)")
```
