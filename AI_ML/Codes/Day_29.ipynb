# Simple Pipeline for Logistic Regression

1. Load Dataset
2. train_test_split
3. scaling
4. model training
5. model evaluation
   

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
data = load_breast_cancer()
data
```

```python
df = pd.DataFrame(data.data, columns = data.feature_names)
df.head()
```

```python
x = df
y = data.target

print(f"Shape of x is : {x.shape}")
print(f"Shape of y is : {y.shape}")
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify = y)

print(f"Shape of x_train is : {x_train.shape}")
print(f"Shape of x_test is : {x_test.shape}")
print(f"Shape of y_train is : {y_train.shape}")
print(f"Shape of y_test is : {y_test.shape}")
```

```python
scaler = StandardScaler()
```

```python
x_train_scaler = scaler.fit_transform(x_train)
x_train_scaler
```

```python
x_test_scaler = scaler.fit_transform(x_test)
x_test_scaler
```

```python
model = LogisticRegression()
```

```python
model.fit(x_train_scaler, y_train)
```

```python
y_pred = model.predict(x_test_scaler)
print(f" accuracy of model : {accuracy_score(y_test,y_pred)}")
```

```python
print(f"Confusion matrix of model is : \n {confusion_matrix(y_test, y_pred)}")
```

```python
print(f"Classification report of model is : \n {classification_report(y_test, y_pred, target_names = data.target_names)}")
```
