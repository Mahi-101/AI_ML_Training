# EDA 

## Univariante

```python
tip = sns.load_dataset("tips")
tip.head()
```

```python
tip.shape
```

## Univariante

```python
plt.figure(figsize= (8,6))
sns.histplot(tip['total_bill'], kde = True, color = 'b')
plt.title("Univariate: Distribution of Total Bills")
plt.xlabel("Total Bills ($)")
plt.ylabel("Frequency")
plt.show()
```

```python
print(tip['day'].value_counts())

plt.figure(figsize= (8,6))
sns.countplot(data = tip, x = 'day', palette = 'Set2')
plt.title("Univariate : Order Frequency by Day")
plt.xlabel("Day of Weeks")
plt.ylabel("Count")
plt.show()
```

## Bivariate 

```python
plt.figure(figsize= (10,7))
sns.regplot(data = tip, x = 'total_bill', y = 'tip', scatter_kws = {'alpha' : 0.9}, color = 'm', marker = 'o')
plt.title("Bivariate: Total Bill VS. Tip Amount")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.show()

corr1 = tip['total_bill'].corr(tip['tip'])
print(f"Correlation between Total Bill VS Tips : {corr1:.2f}")
```

## Categorical VS Numerical

```python
plt.figure(figsize= (10,7))
sns.boxplot(data = tip, x = 'day', y = 'total_bill', palette = 'Set3')
plt.title("Bivariate: Total Bill by Day of Week")
plt.xlabel("Day")
plt.ylabel("Total Bill ($)")
plt.show()
```

```python
plt.figure(figsize= (10,7))
sns.scatterplot(data = tip, x = 'total_bill', y = 'tip', hue = 'smoker', style = 'smoker', s = 70)
plt.title("Multivariate: Total bill vs Tib by smoker status")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.show()
```

```python
tip_num_col = tip.select_dtypes(include = ['int64', 'float64'])
corr_mat = tip_num_col.corr()
```

```python
plt.figure(figsize= (10,7))
sns.heatmap(corr_mat, annot = True, cmap = 'coolwarm', vmin = -1, vmax = 1)
plt.title("Multivariate: Correlation Matrix Heatmap")
plt.show()
```

# Train Test Split

```python
titanic = sns.load_dataset("titanic")
titanic.head()
```

```python
titanic.shape
```

```python
titanic.isnull().sum()
```

```python
null_perc = titanic.isnull().sum()/titanic.shape[0] * 100
null_perc
```

```python
titanic_1 = titanic[['survived', 'pclass', 'age', 'parch']]
titanic_2 = titanic_1.fillna(titanic_1.mean())
titanic_2.head()
```

```python
x = titanic_2.drop('survived', axis = 1)
y = titanic_2['survived']

print('shape of x :', x.shape)
print('shape of y :', y.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3)
print("Shape of x_train = ", x_train.shape)
print("Shape of x_test = ", x_test.shape)
print("Shape of y_train = ", y_train.shape)
print("Shape of y_test = ", y_test.shape)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
print("Shape of x_train = ", x_train.shape)
print("Shape of x_test = ", x_test.shape)
print("Shape of y_train = ", y_train.shape)
print("Shape of y_test = ", y_test.shape)
```
