### Performing np.char functions (np.char --> String Function of numpy)
---

```python
name = 'Mahipal Amrutlal Charaniya'
```

```python
a1 = np.array(['mac', 'abc', 'sac', 'cac'])
a2 = np.array([1,2,3,4])
a3 = np.array(['a', 'b', '3', '5'])
a4 = np.char.array(['mac', 5])
```

```python
print(np.char.isalpha(a3))  # isalnum --> check the array or string contain the alphabets and numeric if yes than true, else false.
# others :- isnumeric, isdecimal, isdigit, islower, istitle, isupper
```

```python
print(np.char.add(a1, a3))
```

```python
print(np.char.upper(name))
```

```python
print(np.char.capitalize(name))
```

```python
print(np.char.asarray(name))
```

```python
print(np.char.center(name, 50, fillchar='*'))
```

```python
print(np.char.compare_chararrays(a1, a3,'==', True))
```

```python
print(np.char.count(a1, 'm'))
```

```python
print(np.char.endswith(a1, 'c'))
```

```python
print(np.char.equal(a1,a3))
```

```python
print(np.char.find(a1, 'm'))
```

```python
print(np.char.greater_equal(a1, a3)) # equal, greater, greater_equal, less, less_equal
```

```python
print(np.char.isnumeric(a4))
```

```python
print(np.char.join(a3, a1))
```

```python
print(np.char.ljust(a1, width=2))
```

```python
print(np.char.lstrip(a1, 'c'))
```

```python
print(np.char.multiply(a1, a2))
```

```python
print(np.char.partition(name, 'p'))
```

```python
print(np.char.replace(a1, 'a', 'A'))
```

```python
print(np.char.rfind(a1, 'a')) #  --> return the index on the finding value
```

```python
print(np.char.str_len(a1))
```
