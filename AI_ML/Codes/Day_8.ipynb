```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from matplotlib.pyplot import style
```

### Continue with Matplotlib 

```python
company = ['ABC', 'MAC', 'SAC', 'CAC', 'AAC', 'HAC', 'AA']
values1 = [100,90,80,70,60,80,75]
values2 = [10,40,50,30,10,20,65]
values3 = [75,60,40,80,50,90,55]

plt.bar(company, values1, color = 'm', edgecolor = 'black', width = 0.7)   # --> use to show the strength of any data. 
plt.title("Company Strength", fontsize = 20, color = 'b')
plt.xlabel("Companies")
plt.ylabel("Strength")
```

```python
# alpha  ---> is use to change the density of the bars means transparency of the bars color

plt.bar(company, values2, color = 'b', edgecolor = 'black', width = 0.7, align = 'center', linewidth = 2, alpha = 0.2)   # --> use to show the strength of any data. 
plt.title("Company Strength", fontsize = 20, color = 'b')
plt.xlabel("Companies")
plt.ylabel("Strength")
```

**NOTE :-** In simple bar graph we can use the width function to change the width of the bars, but in barh(Horizontal bar graph) we have to use height parameter to change the width of the bars.

```python
plt.barh(company, values3, color = 'y', edgecolor = 'black', linestyle = ':')   # --> use to show the strength of any data. 
plt.title("Company Strength", fontsize = 20, color = 'b')
plt.xlabel("Companies")
plt.ylabel("Strength")
```

```python
# visible ---> visible is use to hide the chart if visible is "False" chart is not visible and if "True" chart is visible.

plt.barh(company, values3, color = 'w', edgecolor = 'black', linestyle = ':', visible = False)   # --> use to show the strength of any data. 
plt.title("Company Strength", fontsize = 20, color = 'b')
plt.xlabel("Companies")
plt.ylabel("Strength")
plt.legend(['Company'])
```

```python
# 

style.use('ggplot')
plt.figure(figsize = (16,9))
plt.bar(company, values1, color = 'm', edgecolor = 'black', width = 0.7)   # --> use to show the strength of any data. 
plt.title("Company Strength", fontsize = 20, color = 'b')
plt.xlabel("Companies")
plt.ylabel("Strength")
plt.show()
```

**NOTE:-** When we are plotting two bars in one chart, like overlapping each other, then you have to pass the large values data first and then give the small values data.

```python
#  ---> this show the overlap of the bar graph.

plt.figure(figsize = (16,9))
plt.bar(company, values1, color = 'b', edgecolor = 'black', width = 0.7, label = " Values 2", alpha = 0.5, linestyle = ':', linewidth = 1) 
plt.bar(company, values2, color = 'm', edgecolor = 'black', width = 0.5, label = " Values 1", alpha = 0.7, linestyle = ':', linewidth = 1)   # --> use to show the strength of any data. 
plt.title("Company Strength", fontsize = 20, color = 'b')
plt.xlabel("Companies")
plt.ylabel("Strength")
plt.legend()
```

```python
# here we are implementing group bar chart. here i have also group the line chart.

company = np.arange(len(company))  # ---> this is use to not show the company names and instead of that it will show the length of the company means it show the number. 
# first of all len() give the length of the company and pass to np.arange that convert it to an array of size len()-1.

plt.figure(figsize = (12,6))
width = 0.3
plt.bar(company, values1, width, color = 'b', edgecolor = 'black', label = " Values 2", alpha = 0.5) 
plt.bar(company + width, values2, width, color = 'm', edgecolor = 'black', label = " Values 1", alpha = 0.4, linestyle = ':') 

plt.plot(company, values1, 'ob--')
plt.plot(company+width, values2, 'om--')

# --> use to show the strength of any data. 
plt.grid(visible = False)
style.use('default')
plt.title("Company Strength", fontsize = 20, color = 'b')
plt.xlabel("Companies")
plt.ylabel("Strength")
plt.legend()
```

```python


plt.figure(figsize = (12,6))
width = 0.3
plt.bar(company, values1, width, color = 'b', edgecolor = 'black', label = " Values 2", alpha = 0.5) 
plt.bar(company + width, values2, width, color = 'm', edgecolor = 'black', label = " Values 1", alpha = 0.4, linestyle = ':') 

plt.xticks(company+(width/2), ['ABC', 'MAC', 'SAC', 'CAC', 'AAC', 'HAC', 'AA'], rotation = 20)  # ---> this function is useful to rotated the values of x axis and y axis.

plt.plot(company, values1, 'ob--')
plt.plot(company+width, values2, 'om--')

# --> use to show the strength of any data. 
plt.grid(visible = False)
style.use('default')
plt.title("Company Strength", fontsize = 20, color = 'b')
plt.xlabel("Companies")
plt.ylabel("Strength")
plt.legend()
```

```python
#here we have implement the group bar graph with three different data entries.

plt.figure(figsize = (12,6))
width = 0.3
plt.bar(company, values1, width, color = 'b', edgecolor = 'black', label = " Values 2", alpha = 0.5) 
plt.bar(company + width, values2, width, color = 'm', edgecolor = 'black', label = " Values 1", alpha = 0.4,) 
plt.bar(company + width + width, values3, width, color = 'y', edgecolor = 'black', label = " Values 1", alpha = 0.4,)

plt.xticks(company+(width), ['ABC', 'MAC', 'SAC', 'CAC', 'AAC', 'HAC', 'AA'], rotation = 20)  # ---> this function is useful to rotated the values of x axis and y axis.

plt.plot(company, values1, 'ob--')
plt.plot(company+width, values2, 'om--')
plt.plot(company+width+width, values3, 'oy--')

# --> use to show the strength of any data. 
plt.grid(visible = False)
style.use('default')
plt.title("Company Strength", fontsize = 20, color = 'b')
plt.xlabel("Companies")
plt.ylabel("Strength")
plt.legend()
```

## Plotting Histogram
---

```python
import random
```

```python
# random.randint(start, end, total_number_to_generate)  here we are give a range with start and end and total number we want to generate. 
# this function will generate the randvalues.

ml_student_age = np.random.randint(10, 40, (100))
ai_student_age = np.random.randint(14, 30, (100))
```

```python
ml_student_age
```

```python
ai_student_age
```

```python
plt.hist(ml_student_age)
plt.show()
```

```python
plt.hist(ml_student_age, width = 2)
plt.title("ML Students Age")
plt.xlabel("Student age")
plt.ylabel("No. of students")
plt.show()
```

```python
# bins ---> bins is use to make the custom range for the histogram.

bins = [10,15,20,25,30,35,40]

plt.hist(ml_student_age, bins, rwidth = 0.7)   # rwidth ---> use to make the gap between the bars. small--> more gap, large--> less gap
plt.title("ML Students Age")
plt.xlabel("Student age")
plt.ylabel("No. of students")
plt.show()
```

```python
bins = [10,15,20,25,30,35,40]

plt.hist(ml_student_age, bins, rwidth = 0.7, histtype = 'step')   # rwidth ---> use to make the gap between the bars. small--> more gap, large--> less gap
plt.title("ML Students Age")
plt.xlabel("Student age")
plt.ylabel("No. of students")
plt.show()
```

```python
bins = [10,15,20,25,30,35,40]

plt.hist(ml_student_age, bins, rwidth = 0.7, histtype = 'barstacked')   # rwidth ---> use to make the gap between the bars. small--> more gap, large--> less gap
plt.title("ML Students Age")
plt.xlabel("Student age")
plt.ylabel("No. of students")
plt.show()
```

```python
bins = [10,15,20,25,30,35,40]

plt.hist(ml_student_age, bins, rwidth = 0.7, histtype = 'barstacked', align = 'left')   # rwidth ---> use to make the gap between the bars. small--> more gap, large--> less gap
plt.title("ML Students Age")
plt.xlabel("Student age")
plt.ylabel("No. of students")
plt.show()
```

```python
# orientation --> this function is us to create the chart horizontal and vertical.

bins = [10,15,20,25,30,35,40]

plt.hist(ml_student_age, bins, rwidth = 0.7, histtype = 'barstacked', align = 'left', orientation = 'horizontal')   # rwidth ---> use to make the gap between the bars. small--> more gap, large--> less gap
plt.title("ML Students Age")
plt.xlabel("Student age")
plt.ylabel("No. of students")
plt.show()
```

```python
bins = [10,15,20,25,30,35,40]

plt.hist(ml_student_age, bins, rwidth = 0.7, histtype = 'barstacked', align = 'left', orientation = 'horizontal', color = 'm')   # rwidth ---> use to make the gap between the bars. small--> more gap, large--> less gap
plt.title("ML Students Age")
plt.xlabel("Student age")
plt.ylabel("No. of students")
plt.show()
```

```python
bins = [10,15,20,25,30,35,40]

plt.hist(ml_student_age, bins, rwidth = 0.7, histtype = 'barstacked', align = 'left', orientation = 'horizontal', color = 'm', label = 'chart')   # rwidth ---> use to make the gap between the bars. small--> more gap, large--> less gap
plt.title("ML Students Age")
plt.xlabel("Student age")
plt.ylabel("No. of students")
plt.legend()
plt.show()
```

```python
# here we are merging to histogram chart.

plt.hist(ml_student_age, bins, rwidth = 0.7, histtype = 'barstacked', align = 'left', orientation = 'horizontal', color = 'm', label = 'chart') 
plt.hist(ai_student_age, bins, rwidth = 0.7, histtype = 'barstacked', align = 'left', orientation = 'vertical', color = 'r', label = 'chart') # rwidth ---> use to make the gap between the bars. small--> more gap, large--> less gap
plt.title("ML Students Age")
plt.xlabel("Student age")
plt.ylabel("No. of students")
plt.legend()
plt.show()
```
