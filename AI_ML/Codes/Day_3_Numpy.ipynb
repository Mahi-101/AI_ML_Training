---
### Use of Arange Function
---


```python
arange_arr = np.arange(-5, 5, 1, dtype= np.int64) # negative to postive no. with use of arange function.
print(arange_arr)

arange_arr2 = np.arange(-5, 5, 2, dtype= np.int64) # odd number from -5 to 5 using arange function.
print(arange_arr2)

arange_arr3 = np.arange(-4, 5, 2, dtype= np.int64)  # even number from -4 to 4 using arange function.
print(arange_arr3)
```

```python
# using np.empty funciton
empty_arr = np.empty([3,3], dtype= np.int64)
print(empty_arr)
```

```python
# using np.full function
full_arr = np.full([2,2], 5)
print(full_arr)
```

```python
# use of reshape() --> change the shape of array

reshape_arr = np.reshape(a1, (2,5))
print(reshape_arr)
```

```python
# use of np.linspace() ---> take start, end, and no. of values in array and return the array

line_arr = np.linspace(1, 10, 6, dtype= np.int64)
print(line_arr)
```

```python
# use of np.reshape with np.linspace

line_arr = np.reshape((np.linspace(1, 10, 9, dtype= np.int64)), (3,3))
print(line_arr)

line_arr2 = np.reshape((np.linspace(1, 20, 9, dtype= np.int64)), (3,3))
print(line_arr2)
```

```python
# use of np.ravel ---> make the n-dimension

ravel_arr = np.ravel(a2, 'F')
print(ravel_arr)
```

```python
# use of add, subtract, multiply, divide and dot product

add = np.add(line_arr, line_arr2)
print(add, "\n")

sub = np.subtract(line_arr, line_arr2)
print(sub, "\n")

mul = np.multiply(line_arr, line_arr2)
print(mul, "\n")

div = np.divide(line_arr, line_arr2)
print(div, "\n")

dot = np.dot(line_arr, line_arr2)
print(dot, "\n")
```

```python
a = 'vijay'
b = '-'
b.join(a)
```

```python
# use of np.hstack() and np.vstack()

harr = np.hstack((line_arr, line_arr2))
print("Hstack ", harr, "\n")

varr = np.vstack((line_arr, line_arr2))
print("Vstack ", varr)
```

```python
# use of max, min, argmin, argmax, and mean

max = np.argmax(line_arr)
print("Max : ", np.max(line_arr))
print("Max index : ", max, "\n")

min = np.argmin(line_arr)
print("Min : ", np.min(line_arr))
print("Min index : ", min, "\n")

mean = np.mean(line_arr)
print("Mean : ", mean)
```
