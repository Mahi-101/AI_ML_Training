## Seaborn Introduction

```python
import seaborn as sns
```

```python
# ALL built in dataset in seaborn library

sns.get_dataset_names()
```

```python
# Loading the dataset

data = sns.load_dataset('tips')
data
```

```python
sns.lineplot(x = data['total_bill'], y = data['tip'])
```

```python
sns.lineplot(x = data['day'], y = data['total_bill'], color = 'm')
```

```python
sns.lineplot(x = data['smoker'], y = data['total_bill'], color = 'g')
```

```python
sns.lineplot(x = data['sex'], y = data['total_bill'], color = 'c')
```

```python
sns.lineplot(x = data['size'], y = data['total_bill'], color = 'pink')
```

```python
sns.lineplot(x = data['sex'], y = data['tip'], color = 'orange')
```

```python
sns.lineplot(x = data['time'], y = data['total_bill'], color = 'purple')
```

```python
sns.lineplot(x = data['day'], y = data['smoker'], hue = data['sex'], color = 'olive')
```

```python
car = sns.load_dataset('car_crashes')
car
```

```python
sns.lineplot(x = 'speeding', y = 'total', data = car )
```

```python
sns.lineplot(x = 'alcohol', y = 'total', data = car )
```

```python
sns.distplot(car['speeding'], hist = True,)
```

```python
sns.histplot(car['total'], bins = 10)
```

```python
sns.distplot(car['alcohol'], hist = True, kde = True, rug = True, vertical = True)
```
