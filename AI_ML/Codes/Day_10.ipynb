```python
import seaborn as sns
import matplotlib.pyplot as plt
```

## Scatter Plot 

```python
titanic = sns.load_dataset('titanic')
titanic
```

```python
sns.scatterplot(data = titanic, x = 'age', y = 'fare')
```

```python
# hue ---> by using hue parameter it will differentiate the point based on given categories and apply color so it will easy to visualize.
# style ---> by using the style parameter it will give different-different shapes to each categories of the given column(feature).

sns.scatterplot(data = titanic, x = 'age', y = 'fare', hue = 'sex', style = 'who')
```

```python
# size ---> by using the size parameter it will give the size to all shapes in decreasing order. 

sns.scatterplot(data = titanic, x = 'age', y = 'fare', hue = 'sex', style = 'who', size = 'who')
```

```python
# sizes ---> by the use of this it will make the shape larger in terms of there size.

sns.scatterplot(data = titanic, x = 'age', y = 'fare', hue = 'sex', style = 'who', size = 'who', sizes = (100,100))
```

```python
plt.figure(figsize = (10,7))
sns.scatterplot(data = titanic, x = 'age', y = 'fare', hue = 'sex', style = 'who', size = 'who', sizes = (100,100))
```

```python
# Analyzing the relation between farre and who and alive. most alive are womens and child in compare to men. And some general class women & child are dead only.


plt.figure(figsize = (10,7))
sns.scatterplot(data = titanic, x = 'who', y = 'fare', hue = 'alive', style = 'who', size = 'who', sizes = (100,100))
```

```python
globalwarming = pd.read_csv("E:/AI_ML_Training/AI_ML/raw_materials/Who_is_responsible_for_global_warming.csv")
globalwarming
```

```python
globalwarming.drop(columns = ['Country Code', 'Indicator Name', 'Indicator Code'], axis = 1)
```

```python
globalwarming = globalwarming.drop(columns = ['Country Code', 'Indicator Name', 'Indicator Code'], axis = 1).set_index('Country Name')
globalwarming
```

```python
#  cmap ---> crest, Blues, coolwarm, magma, viridis, YlGnBu, RdYlGn

sns.heatmap(globalwarming, cmap = 'coolwarm')  
```

```python
plt.figure(figsize = (14,7))
sns.heatmap(globalwarming)
```

```python
# vmin, vmax ---> use to define the cbar(range of values bar) range.

plt.figure(figsize = (14,7))
sns.heatmap(globalwarming, vmin = 0, vmax = 20)
```

```python
plt.figure(figsize = (14,7))
sns.heatmap(globalwarming, annot = True, cmap = 'Blues')
```

```python
# annot_kws ---> use to gives the parameter for the annot values visuals.

plt.figure(figsize = (14,7))
annot_kws = {
    'fontsize' : 10,
    'fontstyle' : 'italic',
    'color' : 'r',
    'alpha' : 0.8,
    'rotation' : 'vertical',
    'verticalalignment' : 'center',
    'backgroundcolor' : 'w'
}
sns.heatmap(globalwarming, cmap = 'Blues', annot = True, annot_kws = annot_kws)
```

```python
# linewidth ---> use to increase the width of the heatmap
# linecolor ---> use to chnage the color of the lines.
# k ---> incidate the black color

plt.figure(figsize = (14,7))
sns.heatmap(globalwarming, annot = True, cmap = 'Blues', linewidth = 2, linecolor = 'k')
```

```python
# cbar ---> use to hide the vertical bar that shows the values or we can say range of values with color.

plt.figure(figsize = (14,7))
sns.heatmap(globalwarming, annot = True, cmap = 'Blues', cbar = False)
```

```python
# xticklabels, yticklables ---> use to remove the x-axis label and y-axis labels.

plt.figure(figsize = (14,7))
sns.heatmap(globalwarming, annot = True, cmap = 'Blues', cbar = False, xticklabels = False, yticklabels = False)
```
