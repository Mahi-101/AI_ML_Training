# Pipeline 

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

```python
train = pd.read_csv(r"E:\AI_ML_Training\AI_ML\raw_materials\Day_11\train.csv")
test = pd.read_csv(r"E:\AI_ML_Training\AI_ML\raw_materials\Day_11\test.csv")
print("shape of train dataset", train.shape)
print("shape of test dataset", test.shape)
```

```python
x_train = train.drop(columns = ["SalePrice"], axis = 1)
y_train = train['SalePrice']
x_test = test.copy()
print("shape of x_train dataset", x_train.shape)
print("shape of y_train dataset", y_train.shape)
print("shape of x_test dataset", x_test.shape)
```

## Missing Value imputation

```python
isnull_sum = x_train.isnull().sum()
isnull_sum
```

### Numerical Value Accessing

```python
train_num_var = x_train.select_dtypes(include = ['int64', 'float64']).columns
train_num_var
```

```python
train_num_var_na = [var for var in train_num_var if isnull_sum[var]>0]
train_num_var_na
```

### Categorical Values Accessing

```python
train_cat_var = x_train.select_dtypes(include = ['O']).columns
train_cat_var
```

```python
train_cat_var_na = [var for var in train_cat_var if isnull_sum[var]>0]
train_cat_var_na
```

```python
train_var_mean = ['LotFrontage']
train_var_median = ['MasVnrArea', 'GarageYrBlt']
train_var_mode = ['Alley',
 'MasVnrType',
 'BsmtQual',
 'BsmtCond',
 'BsmtExposure',
 'BsmtFinType1',
 'BsmtFinType2',
 'Electrical',
 'FireplaceQu']

train_var_constant_missing = ['GarageType',
 'GarageFinish',
 'GarageQual',
 'GarageCond',
 'PoolQC',
 'Fence',
 'MiscFeature']
```

## Pipeline

```python
train_var_mean_imputer = Pipeline(steps = [('imputer', SimpleImputer(strategy = 'mean'))])
train_var_medain_imputer = Pipeline(steps = [('imputer', SimpleImputer(strategy = 'median'))])
train_var_mode_imputer = Pipeline(steps = [('imputer', SimpleImputer(strategy = 'most_frequent'))])
train_var_constant_missing_imputer = Pipeline(steps = [('imputer', SimpleImputer(strategy = 'constant', fill_value = 'missing'))])
```

```python
train_var_mean_imputer
```

```python
train_var_medain_imputer
```

### Columns Transformer

```python
preprocessor = ColumnTransformer(transformers = [('mean_imputer', train_var_mean_imputer, train_var_mean),
                                                ('median_imputer', train_var_medain_imputer, train_var_median),
                                                ('mode_imputer', train_var_mode_imputer, train_var_mode),
                                                ('constant_missing', train_var_constant_missing_imputer, train_var_constant_missing)])
```

### applying strategy of dataset

```python
preprocessor.fit(x_train)
```

```python
print(preprocessor.fit(x_train))
```

```python
preprocessor.transform
```

### Accessing Clean Dataset

```python
x_train_clean = preprocessor.transform(x_train)
x_train_clean
```

```python
x_test_clean = preprocessor.transform(x_test)
x_test_clean
```

```python
x_train_clean_df = pd.DataFrame(x_train_clean, columns = train_var_mean + train_var_median + train_var_mode + train_var_constant_missing)
x_train_clean_df.head()
```
