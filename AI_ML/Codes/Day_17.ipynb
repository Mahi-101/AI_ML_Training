# Encoding 

## Label Encoding

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

df = pd.DataFrame({
    "Gender":["Male","Female","Male","Female"]
})

df
```

```python
encoder = LabelEncoder()

df["Gender"] = encoder.fit_transform(df["Gender"])

print(df)
```

## One Hot Encoding

```python
import pandas as pd

df = pd.DataFrame({
    "City":["Pune","Delhi","Mumbai","Pune"]
})

df
```

```python
## we are not using dummy because it work only on binary data and low data, for multiclass and order data we have to use the one hot encoding.

encoded = pd.get_dummies(df,
                         columns=["City"],
                         dtype=int)

print(encoded)
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False)

result = encoder.fit_transform(df[["City"]])

print(result)
```

## Ordinal Encoding

```python
from sklearn.preprocessing import OrdinalEncoder
import pandas as pd

df = pd.DataFrame({
    "Education":[
        "Graduate",
        "PhD",
        "High School",
        "Post Graduate"
    ]
})

df
```

```python
encoder = OrdinalEncoder(categories=[[
    "High School",
    "Graduate",
    "Post Graduate",
    "PhD"
]])

df["Education"] = encoder.fit_transform(df[["Education"]])

print(df)
```
