### 60. apply() to Classify Salary

```python
def classify_salary(salary):
    if salary < 50000:
        return "Low"
    elif salary < 80000:
        return "Medium"
    else:
        return "High"

df["SalaryLevel"] = df["Salary"].apply(classify_salary)
```

```python
df["SalaryLevel"] = df["Salary"].apply(
    lambda x: "High"
    if x >= 80000
    else "Medium"
    if x >= 50000
    else "Low"
)
```

### 62. Use groupby() with Multiple Aggregations

```python
result = df.groupby("Department").agg({
    "Salary": ["min", "max", "mean", "sum"],
    "Age": ["min", "max", "mean"],
    "Rating": ["mean", "max"]
})

print(result)
```
