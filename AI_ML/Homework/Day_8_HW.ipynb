## Q.1) Why do we use EDA (Exploratory Data Analysis)?
## Ans --> 
EDA is the process of exploring and understanding data before building a machine learning model.

**Why do we use it?**
* Understand the data structure.
* Find missing values.
* Detect outliers.
* Identify relationships between variables.
* Check data distribution.
* Select useful features.
* Decide preprocessing steps.

**Example**
Suppose you're building a model to predict whether a customer will cancel a gym membership.

**EDA might reveal:**
* Most customers who cancel visit the gym less than 3 times/month.
* 20% of customers have missing age values.
* Membership duration is highly correlated with cancellation.
  
This information helps prepare better data before training.

## Q.3) What is an ROC curve? How do we measure it?
## Ans -->
ROC stands for Receiver Operating Characteristic Curve. It evaluates the performance of a binary classifier.

It plots: **True Positive Rate (TPR) vs False Positive Rate (FPR)**

**Formulas**

True Positive Rate: **TPR = TP / (TP + FN)**

False Positive Rate: **FPR = FP / (FP + TN)**
    
**Measurement**

We use AUC (Area Under Curve).
| AUC | Meaning      |
| --- | ------------ |
| 1.0 | Perfect      |
| 0.9 | Excellent    |
| 0.8 | Good         |
| 0.7 | Average      |
| 0.5 | Random Guess |

**Example :**
Suppose a hospital AI predicts whether a patient has pneumonia. Changing the prediction threshold changes TPR and FPR. ROC shows which threshold gives the best balance.

## Q.5) What is Standard Deviation?
## Ans -->
**Definition:** Standard deviation measures how spread out data is from the mean.

**Low SD**
→ Data points are close together.

**High SD**
→ Data points are spread out.

**Formula**
$$
\sigma = \sqrt{\frac{1}{N}\sum_{i=1}^{N}(x_i-\mu)^2}
$$

Where:

- $\sigma$ = Population Standard Deviation
- $N$ = Total number of observations
- $x_i$ = Each data value
- $\mu$ = Population mean

## Q.7) What are Variance and Covariance?
## Ans -->
### **Variance:** Measures spread of one variable.

**Formula**
$$
\sigma^2 = \frac{1}{N}\sum_{i=1}^{N}(x_i-\mu)^2
$$

Where:

- $\sigma^2$ = Population Variance
- $N$ = Total number of observations
- $x_i$ = Each data value
- $\mu$ = Population mean

Higher variance means greater spread.

### **Covariance:** Measures how two variables move together.

**Positive covariance:** Both increase together.

**Negative covariance:** One increases while the other decreases.

**Near zero:** No relationship.

**Example** During summer:

Temperature ↑

Ice cream sales ↑

Positive covariance.

## Q.9) What is Classification Matrics?
## Ans -->
It evaluates classification models.

|                 | Predicted Positive | Predicted Negative |
| --------------- | ------------------ | ------------------ |
| Actual Positive | TP                 | FN                 |
| Actual Negative | FP                 | TN                 |

**From this we calculate:**
**Accuracy: (TP+TN)/(Total)**

**Precision: TP/(TP+FP)**

**Recall: TP/(TP+FN)**

**F1 Score: (2 × Precision × Recall)/(Precision + Recall)**

**Example:** A model identifies defective products on a production line.

The confusion matrix tells:

* Correct defect detections
* Missed defects
* False alarms

## Q.10) How to calculate Learninng Rate?
## Ans -->
The learning rate (η) is a hyperparameter, not something usually calculated directly from the data. It controls how much the model updates its weights during training.

**Weight update rule:** New Weight = Old Weight − (η × Gradient)

where:
* η = learning rate
* Gradient = slope of the loss function

**Example** Suppose:
* Old weight = 2.5
* Gradient = 0.8
* Learning rate = 0.1

**Then:** New Weight = 2.5 − (0.1 × 0.8) = 2.42

**Choosing a learning rate**
* Too high (e.g., 1.0): Training may overshoot the optimum and fail to converge.
* Too low (e.g., 0.00001): Training becomes very slow.
* Common starting values: 0.1, 0.01, 0.001 (depending on the optimizer and model).

Many optimizers (like Adam) adapt the effective step size automatically.

## Q.11) What are Overfitting and Underfitting?
## Ans -->
### Overfitting

Overfitting is a machine learning problem in which a model learns not only the important patterns in the training data but also the noise, outliers, and random fluctuations. As a result, the model performs very well on the training data but poorly on new, unseen data because it fails to generalize.

Training Accuracy:
99%

Testing Accuracy:
70%

Poor generalization.

**Example:**
A plant disease model learns the background color of training images instead of the leaf patterns. It performs well on training images but poorly on new photos.

### Underfitting

Underfitting is a machine learning problem in which a model is too simple to learn the underlying patterns in the training data. As a result, it performs poorly on both the training data and unseen (testing) data because it cannot capture the relationships between the input features and the target.

Training Accuracy:
60%

Testing Accuracy:
58%

Cannot learn patterns.

**Example:**
Using a straight-line model to predict the curved growth pattern of a crop over time.

## Q.12) What is Encoding and Decoding?
## Ans --> 
### Encoding

Encoding is the process of converting data from its original form into a numerical or machine-readable representation so that machine learning models or neural networks can process it effectively.

**Examples:**
* One-Hot Encoding
* Label Encoding
* Word Embeddings

**Example:** Fruit = (Apple, Orange, Banana)

Label Encoding = (Apple → 0, Orange → 1, Banana → 2 )

### Decoding

Decoding is the process of converting encoded numerical or internal representations back into a human-readable or meaningful form.

**Example:**
A translation model encodes an English sentence into internal numerical features and decodes those features into a French sentence.

## Q.13) What are Forward Propogation and Backward Propogation?
## Ans --> 
### Forward Propagation

Forward propagation is the process in which the input data passes through the neural network from the input layer, through one or more hidden layers, to the output layer. During this process, each neuron computes a weighted sum of its inputs, applies an activation function, and passes the result to the next layer to generate the model's prediction.

**Data moves:** Input --> Hidden Layers --> Output --> Prediction is made.

### Backward Propagation

Backward propagation (Backpropagation) is the process of propagating the prediction error from the output layer back through the hidden layers to the input layer. It computes the gradient of the loss function with respect to each weight and bias using the chain rule of calculus, allowing the optimizer to update the model parameters and minimize the prediction error.

**Error moves backward:** Loss --> Output Layer --> Hidden Layer --> Input Layer --> Weights are updated using gradients.

## Q.14) Difference bewteen Perceptron and Neuron?
## Ans -->
| Perceptron                                | Neuron                                                         |
| ----------------------------------------- | -------------------------------------------------------------- |
| Simplest artificial neural unit           | General computational unit in neural networks                  |

A neuron in a deep neural network can help recognize handwritten characters with complex shapes.
