# Scalars, Vectors, Matrices, and Tensors

---

# Introduction

Before learning Machine Learning, Deep Learning, or Artificial Intelligence, it is important to understand the mathematical objects used to represent data.

The four basic mathematical objects are:

- Scalar
- Vector
- Matrix
- Tensor

These objects differ based on the number of dimensions they contain.

| Object | Dimensions |
|---------|------------|
| Scalar | 0-D |
| Vector | 1-D |
| Matrix | 2-D |
| Tensor | 3-D or Higher |

---

# 1. Scalar

## Definition

A **Scalar** is a single numerical value.

It has:

- Magnitude
- No direction
- Zero Dimensions (0-D)

Examples include:

- Age = 21
- Temperature = 35°C
- Speed = 60 km/h
- Weight = 70 kg

---

## Mathematical Representation

A scalar is usually represented by lowercase letters.

Example:

\[
a = 10
\]

\[
b = -5
\]

\[
c = 3.14
\]

---

## Examples

Example 1

```
Age = 20
```

Example 3

```
Learning Rate = 0.001
```

---

## Real-Life Examples

- Number of students
- Salary
- Height
- Marks
- Price

---

# 2. Vector

## Definition

A **Vector** is an ordered collection of numbers.

It has

- Magnitude
- Direction (in Mathematics/Physics)
- One Dimension (1-D)

---

## Representation

Vectors are written inside brackets.

Example

\[
v =
\begin{bmatrix}
2\\
4\\
6
\end{bmatrix}
\]

or

\[
v=[2,4,6]
\]

---

## Example

Student Marks

```
[80, 90, 95]
```

RGB Color

```
[255, 128, 0]
```

---

## Size of Vector

If a vector contains **n** elements,

then

```
Size = n
```

Example

```
[2,4,6,8]
```

contains

```
4 elements
```
---

## Applications

- GPS Coordinates
- Feature Vectors
- Embeddings
- Sensor Data
- Audio Signals

---

# 3. Matrix

## Definition

A **Matrix** is a rectangular arrangement of numbers in rows and columns.

A matrix is a **2-Dimensional object**.

---

## Representation

Example

\[
A=
\begin{bmatrix}
1&2&3\\
4&5&6
\end{bmatrix}
\]

This matrix contains

- 2 Rows
- 3 Columns

Therefore

```
Shape = (2,3)
```

---

## Matrix Terminology

Example

```
1 2 3
4 5 6
```

Rows = 2

Columns = 3

---

## Types of Matrices

### Row Matrix

```
[1 2 3]
```

Shape

```
(1,3)
```

---

### Column Matrix

```
1
2
3
```

Shape

```
(3,1)
```

---

### Square Matrix

Rows = Columns

Example

```
1 2
3 4
```

Shape

```
(2,2)
```

---

### Identity Matrix

Diagonal values are 1.

Example

```
1 0 0
0 1 0
0 0 1
```

---

### Zero Matrix

Every element is zero.

Example

```
0 0
0 0
```
---

## Applications

- Image Processing
- Spreadsheet Data
- Neural Networks
- Graph Algorithms
- Recommendation Systems

---

# 4. Tensor

## Definition

A **Tensor** is a generalization of Scalars, Vectors, and Matrices.

A tensor can have

- 3 Dimensions
- 4 Dimensions
- 5 Dimensions
- n Dimensions

---

## Hierarchy

```
Scalar  → 0-D

Vector → 1-D

Matrix → 2-D

Tensor → 3-D or Higher
```

---

## Example of a 3D Tensor

Imagine two matrices stacked together.

```
Matrix 1

1 2
3 4

Matrix 2

5 6
7 8
```

Tensor Shape

```
(2,2,2)
```

---

## Higher-Dimensional Tensor

Example

Video Data

```
Frames × Height × Width × Channels
```

Example Shape

```
(100,224,224,3)
```

Meaning

- 100 Frames
- Height = 224
- Width = 224
- RGB Channels = 3

---

## Applications

- Deep Learning
- Computer Vision
- Natural Language Processing
- Medical Imaging
- Scientific Computing

---
